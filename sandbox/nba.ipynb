# NBA data

```python
import nba_api
import pandas as pd
```

Get Grizzlies team ID

```python
from nba_api.stats.static import teams
from nba_api.stats.endpoints import leaguegamefinder
from datetime import datetime

griz = teams.find_team_by_abbreviation("mem")
griz_games = leaguegamefinder.LeagueGameFinder(team_id_nullable=griz["id"])
df = griz_games.get_data_frames()[0]
df["date_time"] = pd.to_datetime(df["GAME_DATE"])


df[["GAME_ID", "GAME_DATE", "date_time"]]
df["date_time"].dt.year

# df.loc[0]["GAME_DATE"]
# game_date = datetime.strptime(df.loc[0]["GAME_DATE"], "%Y-%m-%d")
# game_date.year >= 2027
```

# Get box score of a game

```python
from nba_api.stats.endpoints import boxscoretraditionalv3

# 1. Grab your target game ID
igame = df.loc[0]["GAME_ID"]
print(f"Fetching data for game ID: {igame}")

# 2. Use the stable historical traditional box score endpoint
box = boxscoretraditionalv3.BoxScoreTraditionalV3(game_id=igame, timeout=30)

# 3. Pull the dataframes (Index 0 is Player Stats, Index 1 is Team Stats)
df_players = box.get_data_frames()[0]
df_teams = box.get_data_frames()[1]

df_teams
```
